## Langkah 1: Siapkan alat dan muat data

```python
import pandas as pd
import matplotlib.pyplot as plt

DATA_PATH = "data/mall_customers.csv"
df = pd.read_csv(DATA_PATH)
```

## Langkah 2: Lihat sekilas isi data

```python
print("Lima baris pertama:")
print(df.head())

print("\nUkuran (baris, kolom):", df.shape)

print("\nTipe data per kolom:")
print(df.dtypes)
```

## Langkah 3: Ringkasan statistik

```python
print("\nRingkasan statistik:")
print(df.describe().round(2))
```

## Langkah 4: Pemeriksaan kualitas dasar

```python
print("\nJumlah nilai kosong per kolom:")
print(df.isna().sum())

print("\nJumlah baris duplikat:", df.duplicated().sum())

print("\nJumlah per Gender:")
print(df["Gender"].value_counts())
```

## Langkah 5: Histogram

```python
features = ["Age", "Annual Income (k$)", "Spending Score (1-100)"]

fig, axes = plt.subplots(1, 3, figsize=(12, 3.5))
for ax, col in zip(axes, features):
    ax.hist(df[col], bins=15, edgecolor="white")
    ax.set_title(col)
    ax.set_ylabel("Jumlah pelanggan")
fig.tight_layout()
plt.show()
```

## Langkah 6: Scatter plot

```python
fig, ax = plt.subplots(figsize=(5.5, 4.5))
ax.scatter(df["Annual Income (k$)"], df["Spending Score (1-100)"], s=18)
ax.set_xlabel("Annual Income (k$)")
ax.set_ylabel("Spending Score (1-100)")
ax.set_title("Pendapatan vs skor belanja")
plt.show()
```
